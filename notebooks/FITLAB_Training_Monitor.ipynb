# Theo dõi huấn luyện GPU
Notebook này chỉ đọc log/checkpoint; không tự khởi động thêm job train. Chạy lại cell để cập nhật. Mô hình: content-aware Transformer two-tower, dùng bộ lớn 80/10/10. Test chưa được đánh giá trong quá trình train.

```python
from pathlib import Path
import json, subprocess
ROOT = Path('/home/coder/iDragonCloud/DA_AI')
RUN = ROOT / 'runs/content_transformer_v1'
log = ROOT / 'runs/content_transformer_v1.log'
if not log.exists():
    log = ROOT / 'runs/smoke_gpu_v1.log'
print('LOG:', log)
if log.exists():
    print(subprocess.run(['tail', '-n', '16', str(log)], capture_output=True, text=True).stdout)
else:
    print('Chưa có log.')
for name in ('best.pt', 'latest.pt', 'completed.json'):
    path = RUN / name
    if path.exists():
        print(name, round(path.stat().st_size / 2**20, 1), 'MiB')
print(subprocess.run(['nvidia-smi', '--query-gpu=name,memory.used,memory.total,utilization.gpu', '--format=csv'], capture_output=True, text=True).stdout)
```

```python
history = RUN / 'history.jsonl'
if history.exists():
    for line in history.read_text().splitlines():
        row = json.loads(line)
        val = row['validation']
        print('Epoch', row['epoch'], '| loss', row['train_loss'], '| overall', val['overall'], '| có lịch sử', val['known_user'])
else:
    print('Chưa hoàn thành epoch đầu tiên.')
```
